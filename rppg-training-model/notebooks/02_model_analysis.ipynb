# Model Analysis

In this notebook, we will analyze the performance of the trained rPPG model. We will visualize the predictions, calculate performance metrics, and perform error analysis.

```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from src.evaluate import evaluate_model
from src.utils import load_data

# Set visualization style
sns.set(style='whitegrid')

# Load the test dataset
test_data = load_data('data/splits/test.csv')

# Evaluate the model
predictions, ground_truth = evaluate_model('checkpoints/best_model.pth', test_data)

# Calculate performance metrics
mae = np.mean(np.abs(predictions - ground_truth))
mse = np.mean((predictions - ground_truth) ** 2)
rmse = np.sqrt(mse)

# Display metrics
print(f'Mean Absolute Error (MAE): {mae:.2f}')
print(f'Mean Squared Error (MSE): {mse:.2f}')
print(f'Root Mean Squared Error (RMSE): {rmse:.2f}')

# Visualize predictions vs ground truth
plt.figure(figsize=(12, 6))
plt.plot(ground_truth, label='Ground Truth', color='blue')
plt.plot(predictions, label='Predictions', color='orange')
plt.title('Model Predictions vs Ground Truth')
plt.xlabel('Samples')
plt.ylabel('Heart Rate (BPM)')
plt.legend()
plt.show()

# Error analysis
errors = predictions - ground_truth
plt.figure(figsize=(12, 6))
sns.histplot(errors, bins=30, kde=True)
plt.title('Error Distribution')
plt.xlabel('Prediction Error (BPM)')
plt.ylabel('Frequency')
plt.axvline(0, color='red', linestyle='--')
plt.show()
```
