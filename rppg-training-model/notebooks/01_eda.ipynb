# Exploratory Data Analysis (EDA) for rPPG Dataset

In this notebook, we will perform exploratory data analysis on the rPPG dataset. We will visualize the data distributions, relationships, and any patterns that may exist in the dataset.

```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visualization style
sns.set(style='whitegrid')
```

```python
# Load the dataset
subject41_data = pd.read_csv('../data/raw/ubfc_rppg/subject41.csv')
subject42_data = pd.read_csv('../data/raw/ubfc_rppg/subject42.csv')
subject46_data = pd.read_csv('../data/raw/ubfc_rppg/subject46.csv')

# Display the first few rows of one of the datasets
subject41_data.head()
```

```python
# Summary statistics
subject41_data.describe()
```

```python
# Visualize the distribution of heart rates
plt.figure(figsize=(10, 6))
sns.histplot(subject41_data['heart_rate'], bins=30, kde=True)
plt.title('Distribution of Heart Rates for Subject 41')
plt.xlabel('Heart Rate (BPM)')
plt.ylabel('Frequency')
plt.show()
```

```python
# Visualize the relationship between heart rate and time
plt.figure(figsize=(12, 6))
plt.plot(subject41_data['time'], subject41_data['heart_rate'], label='Heart Rate', color='b')
plt.title('Heart Rate Over Time for Subject 41')
plt.xlabel('Time')
plt.ylabel('Heart Rate (BPM)')
plt.legend()
plt.show()
```

```python
# Correlation heatmap
plt.figure(figsize=(10, 8))
correlation_matrix = subject41_data.corr()
sns.heatmap(correlation_matrix, annot=True, fmt='.2f', cmap='coolwarm', square=True)
plt.title('Correlation Heatmap for Subject 41')
plt.show()
```

## Conclusion

In this notebook, we performed exploratory data analysis on the rPPG dataset. We visualized the distribution of heart rates, analyzed the relationship between heart rate and time, and examined the correlation between different features. Further analysis can be conducted to derive more insights.
